```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('bus_travel_time_dataset.csv')
print('Shape:', df.shape)
df.head()
```

```python
df = df.drop_duplicates().dropna()
X = df.drop(columns=['Travel_Time_min'])
y = df['Travel_Time_min']
cat_cols = ['Route', 'Weather', 'Traffic_Level']
preprocessor = ColumnTransformer([('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)], remainder='passthrough')
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=29)
model = Pipeline([('preprocessor', preprocessor), ('regressor', RandomForestRegressor(n_estimators=150, random_state=29))])
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('MAE:', round(mean_absolute_error(y_test, pred), 2), 'minutes')
print('RMSE:', round(np.sqrt(mean_squared_error(y_test, pred)), 2), 'minutes')
print('R2:', round(r2_score(y_test, pred), 3))
```

```python
new_trip = pd.DataFrame([{'Route':'Route A','Distance_km':12.5,'Departure_Hour':8,'Day_of_Week':1,'Weather':'Rain','Traffic_Level':'High','Number_of_Stops':4}])
print('Estimated travel time:', round(model.predict(new_trip)[0], 1), 'minutes')
```
